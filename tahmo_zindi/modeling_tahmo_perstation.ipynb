```python
# creating model using random forest
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
```

```python
train.columns
```

```python
train.columns.tolist()
```

```python
X_columns = [ 'precipitation (mm)',
 'relativehumidity (-)',
 'temperature (degrees Celsius)',
 'installation_height',
 'elevation',
 'latitude',
 'longitude']
```

```python
X = train[X_columns]
```

```python
X.head()
```

```python
y = train['radiation (W/m2)']
```

```python
# create simple prediction
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
# create random forest model
model_rf=RandomForestRegressor()
model_rf.fit(X_train,y_train)
```

```python
print(model_rf)
```

```python
from sklearn.metrics import mean_absolute_error, root_mean_squared_error,r2_score
```

```python
import numpy as np
from typing import Union,Dict
```

```python
# create evaluation function
def eval_metrics(y_true:Union[np.ndarray,list],y_pred:Union[np.ndarray,list])->Dict[str,float]:
    # tulis evaluation metrics
    mae = mean_absolute_error(y_true,y_pred)
    rmse = root_mean_squared_error(y_true,y_pred)
    r2 = r2_score(y_true,y_pred)

    y_true = np.array(y_true)
    y_pred=np.array(y_pred)

    mape = np.mean(np.abs((y_true - y_pred) / np.where(y_true == 0, 0.01, y_true))) * 100

    MBE = np.mean(np.abs(y_true - y_pred))

    rmse_mbe = 0.5*rmse+0.5*MBE

    return {'MAE':round(mae,2),
    'RMSE':round(rmse,2),
    'R2':round(r2,2),
    'MAPE':round(mape,2),
    'mbe':round(MBE,2),
    'rmse_mbe':round(rmse_mbe,2)}
```

```python
# save model
y_prediction = model_rf.predict(X_test)
```

```python
model_perf = eval_metrics(y_test,y_prediction)
```
